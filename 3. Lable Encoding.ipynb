Lable encoding 

converting the lables into numeric value
(e.g) ==> name string value (a,a,a,b,b) ==> (0,0,0,1,1)
a ==> 0
b ==> 1 

```python
# import libaries
import pandas as pd
from sklearn.preprocessing import LabelEncoder 
```

Lable Encoding of reast Cancer DataSet

```python
# loading the data from csv file to panda dataFrame
cencer_data =  pd.read_csv("DataSets/breast_cancer_data.csv")
```

```python
cencer_data.head()
```

```python
# find the count of differnt values 
cencer_data["diagnosis"].value_counts()
```

```python
# load the labale encoding 
lable_encode = LabelEncoder()
```

```python
lable =  lable_encode.fit_transform(cencer_data["diagnosis"])
```

```python
# adding the new lable to the data frame
cencer_data["target"] = lable
```

```python
cencer_data.head()
```

```python
cencer_data["target"].value_counts()
```

0 -->  Benign

1 --> Malignant
